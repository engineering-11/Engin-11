## 9. Masking and slicing data

A *mask* is a boolean array — `True` for rows you want to keep, `False` for rows to drop. Pass the mask back to the DataFrame to filter.

```python
# Drop any rows where cpm is non-positive
mask = data['cpm'] > 0
data_cut = data[mask]
display(data_cut)
```

```python
# Filter by date — combine multiple conditions with & (and) / | (or). Parentheses are required.
mask = (data_cut['deviceTime_local'] > '2022-01-01 00:00:00-0800') & \
       (data_cut['deviceTime_local'] < '2023-01-01 00:00:00-0800')
data_2022 = data_cut[mask]
display(data_2022)
```

```python
# Drop columns we don't need by slicing — .loc[rows, cols] takes column-label ranges too
data_cut = data_cut.loc[:, 'deviceTime_local':'cpm']
display(data_cut)
```

## 11. Computing windowed averages

Detector data is often noisy at a single sample but cleaner when averaged over a longer interval. Here's a basic for-loop pattern for computing daily averages.

```python
# One-day window in seconds
interval = 60 * 60 * 24
```

```python
start_time = data_2022['deviceTime_unix'].iloc[0]
end_time   = data_2022['deviceTime_unix'].iloc[-1]

daily_aves = []
t = start_time
while t < end_time:
    mask = (data_2022['deviceTime_unix'] >= t) & (data_2022['deviceTime_unix'] < t + interval)
    chunk = data_2022[mask]
    if len(chunk) > 0:
        daily_aves.append(chunk['cpm'].mean())
    t += interval

print(f'computed {len(daily_aves)} daily averages')
```

```python
plt.plot(daily_aves)
plt.xlabel('day index')
plt.ylabel('mean cpm')
plt.show()
```

## 12. Wrapping logic in a function

When you find yourself running the same calculation more than once, wrap it in a function. That makes the code easier to read, test, and reuse.

```python
def get_ave(data, data_type, interval):
    '''Compute averages of `data_type` over fixed time windows.'''
    start_time = data['deviceTime_unix'].iloc[0]
    end_time   = data['deviceTime_unix'].iloc[-1]
    averages = []
    t = start_time
    while t < end_time:
        mask = (data['deviceTime_unix'] >= t) & (data['deviceTime_unix'] < t + interval)
        chunk = data[mask]
        if len(chunk) > 0:
            averages.append(chunk[data_type].mean())
        t += interval
    return averages
```

```python
# Daily average cpm
daily = get_ave(data_2022, 'cpm', 60 * 60 * 24)
plt.plot(daily)
plt.xlabel('day index')
plt.ylabel('mean cpm')
plt.show()
```

```python
# Hourly average — same function, smaller window
hourly = get_ave(data_2022, 'cpm', 60 * 60)
plt.plot(hourly)
plt.xlabel('hour index')
plt.ylabel('mean cpm')
plt.show()
```
